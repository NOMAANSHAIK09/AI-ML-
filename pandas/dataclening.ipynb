```python
#melt and pivot
df=pd.DataFrame({
    "country":["usa","usa","india","india"],
    "year":[2020,2021,2020,2021],
    "sales":[100,120,90,110],
    "profit":[20,25,18,22]
}
               )
melted_data=df.melt(
    id_vars = ["country","year"],
    value_vars=["sales","profit"],
    var_name = "metrics",
    value_name="values"
)
melted_data
```

```python
orignal = melted_data.pivot(
    index = ["country","year"],
    columns = "metrics",
    values="values"
)
orignal
```
