```python
from openai import OpenAI

client = OpenAI(
    api_key=OPEN_API_KEY,
    base_url="https://openrouter.ai/api/v1"
)
```

```python
class PersonalAssistant:
  def __init__(self):
    print("hi, i am your ai assistant.how can i help u")

  def ans_query(self):
        question=input("Ask me any thing:")
        response = client.chat.completions.create(
            model="openai/gpt-4o-mini",
            messages=[
                {
                    "role": "system",
                    "content": "Act like a helpful assistant."
                },
                {
                    "role": "user",
                    "content": question
                }
            ],
            temperature=0.7,
            max_tokens=512
        )

        print(response.choices[0].message.content.strip())

  def summarize_email(self):
        email_text=input("Enter the email text:")
        prompt=f"Summarize the following email in two-three sentence: {email_text}"
        response = client.chat.completions.create(
            model="openai/gpt-4o-mini",
            messages=[
                {
                    "role": "system",
                    "content": "Act like a expert email assistant."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            temperature=0.3,
            max_tokens=512
        )

        print("\n\nsummary:",response.choices[0].message.content.strip())

```

```python
assistant=PersonalAssistant()
```

```python
assistant.ans_query()
```

```python
assistant.summarize_email()
```

# image API



```python
prompt="A highly detailed futuristic spaceship flying near a beautiful colorful nebula"

result=client.images.generate(
    model="gpt-images-1.5"
    prompt=prompt
)

image_base64=result.data[0].b64_json
image_bytes=base64.b64decode(image_base64)

with open("spaceship.png","wb") as f:
  f.write(image_bytes)
```

Image edit

```python
prompt="Take the reference image and convert the spaceship color to red"
result=client.images.edit(
    model="gpt-images-1.5"
    prompt=prompt
    image=[
        open("spaceship-image.png","rb")
    ]
)

image_base64=result.data[0].b64_json
image_bytes=base64.b64decode(image_base64)

with open("spaceship.png","wb") as f:
  f.write(image_bytes)
```
