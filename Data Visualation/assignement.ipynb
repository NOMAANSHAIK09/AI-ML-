```python
#Compare temp of cities over a week using subplots and loops
```

```python
import matplotlib.pyplot as plt
```

```python
days =["mon", "tue", "wed", "thu", "fri"]

cities = ["new york", "london", "delhi", "tokyo"]

temps=[
[22,23,121,24,25],
[18,19,17,20,21],
[30,32,31,33,34],
[25,26,24,28,23]
]
fig, axes = plt.subplots(2,2)
cityno = 0 
for i in range(2):
    for j in range(2):
        axes[i][j].plot(days,temps[cityno])
        axes[i][j].set_title(cities[cityno])
        axes[i][j].grid(True)
        cityno=cityno+1
fig.suptitle("temp in cities over the week")
plt.tight_layout()
```

```python
#Q5tips•Createapivottableofaveragetip◦Rows:day◦Columns:time•Plotthepivottableasaheatmap•Annotatevalues•Interpretwhichday–timecombinationperformsbest
```

```python
data_p=tips.pivot_table(
    values="tip",
    index="day",
    columns="time",
    aggfunc="mean"
)

data_p
```

```python
sns.heatmap(
    data_p,
    annot=True,
    # cmap="YlGnBu",
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Average Tip by Day and Time")
```
