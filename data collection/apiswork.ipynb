```python
import requests
url="https://cat-fact.herokuapp.com"

res=requests.get(url)

```

```python
print("Status Code:", res.status_code)
print("Response Text:", res.text)

try:
    data = res.json()
    print(data)
except Exception as e:
    print("JSON Error:", e)
```
