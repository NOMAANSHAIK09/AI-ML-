```python
#vectorization and Broadcasting
import numpy as np 

# vectorization
arr = np.array([1,2,3,4,5,6])
arr2=np.array([7,8,9,0,1,2])
print(arr ** 2)
print(arr + arr2)


```

```python
#Broadcasting
print(arr + 5)
```

```python
#normalize

arr2D=np.array([[1,2],[3,4]])
mean=np.mean(arr2D)
stddev=np.std(arr2D)
normalized=(arr2D-mean)/stddev
print(normalized)
```
